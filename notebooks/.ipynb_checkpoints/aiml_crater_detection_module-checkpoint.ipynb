# Function to detect crater on the surface

```python
def aiml_crater_detection(image_name):
    from ultralytics import YOLO
    import cv2
    import matplotlib.pyplot as plt
    
    # Load model
    model = YOLO("new_best.pt")
    
    # Your test image
    image_path = "test_images/" + image_name + ".jpg"
    
    # Load original image
    image = cv2.imread(image_path)
    
    if image is None:
        raise FileNotFoundError(f"Could not find image: {image_path}")
    
    # Run YOLO detection
    results = model.predict(
        source=image_path,
        conf=0.30,
        verbose=False
    )
    
    result = results[0]
    
    # Create annotated image
    annotated = result.plot(
        conf=True,
        labels=True,
        line_width=4
    )
    
    # Convert BGR -> RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    annotated_rgb = annotated[..., ::-1]
    
    # Number of detections
    count = len(result.boxes)
    
    # ============================================================
    # DISPLAY SIDE-BY-SIDE
    # ============================================================
    
    fig, axes = plt.subplots(1, 2, figsize=(7, 4))
    
    # Original image
    axes[0].imshow(image_rgb)
    axes[0].set_title("Input Image")
    axes[0].axis("off")
    
    # YOLO detection result
    axes[1].imshow(annotated_rgb)
    axes[1].set_title(f"AI/ML Crater Detection - {count} craters")
    axes[1].axis("off")
    
    plt.tight_layout()
    plt.show()
    
    # ============================================================
    # PRINT DETECTION RESULTS
    # ============================================================
    
    print("AI/ML Crater Detection")
    print("---------------------")
    print("Detected craters:", count)
    
    for i, box in enumerate(result.boxes):
    
        confidence = float(box.conf[0])
    
        print(
            f"Crater {i+1}: "
            f"confidence = {confidence:.3f}"
        )
```
