# Text to Speech Conversion

# Text to Speech Conversion with saving audio

```python
def speechSaveFile():
    # retreving a module for text to speech conversion
    import gtts

    # importing system module
    import os

    input_text = "I am Amazing"

    # setting speech language
    language = 'en'

    # invoking text to speech conversion function
    # slow = False for faster play of audio

    audio = gtts.gTTS(text=input_text,lang=language,slow=False)

    # saving the audio in a file
    file_name = "intro"+".mp3"
    audio.save(file_name)

    # playing the saved audio
    os.system(file_name)
    
```

# Text to Speech Conversion without saving audio

```python
def voice(text_to_play):
    
    #importing Text to speech module
    import pyttsx3
    
    # creating an object
    eng = pyttsx3.init()
    
    # set the audio speed
    eng.setProperty("rate",130)

    # loading text for speech
    eng.say(text_to_play)

    # to execute audio play
    eng.runAndWait()
    
    
```
